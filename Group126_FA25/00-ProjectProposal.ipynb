## Research Question

We want to test if there are stastically siginificant difference of "review behaviors" between "critical users", "generous users" and "normal users."
- Consider Google Map reveiw data between 2014 and 2024 in California.
- We define "highly-rated restaurants" as restaurants which has at leat 100 reviews whose rating is higher than 4.8 and "harshly-rated restauratns" as restaurants which has at least 100 reviews whose rating is lower than 1.5.
- We consider users who has more than 50 reviews and we define "critical users" as users who rated at least two of such "highly-rated restaurants" 0 or 1 star, "generous users" as users who rated at least two of such "harshly-rated restaurants" 4 or 5 stars, and "normal users" as those who are not "cirical users" or "generous users". 
- We define "review behaviors" as users' average review length (words), average star rating, frequency of reviews per month, and the number of different counties where their reviewd restaurants are located.

## Background and Prior Work

Online reviews play a significant role in consumers' decision-making process and a business's reputation, particularly on Google Maps. Users influence potential customers, search engine positioning, and, in some cases, whether a business thrives through ratings in the form of stars, written commentary, and even pictures they post. Although they pervade, there is still much to be discovered about the reviewers themselves, such as how different users behave across multiple reviews and whether certain users consistently deviate from the majority in recognizable ways. For this project, we will identify "critical," "generous," and "neutral" reviewers based on how their ratings compare to the average rating of the establishments they review. We will also determine whether these categories of reviewers exhibit distinct behavioral patterns.

Previous research has noted that online review distributions are highly polarized, with most users leaving extremely positive or extremely negative reviews, and proportionately fewer in between. This theory confirms that we can successfully categorize and recognize reviewers into at least two categories: critical and generous. Studies such as When Consistency Matters: The Effect of Valence Consistency on Review Helpfulness (Quaschning, Pandelaere & Vermeir, 2014) show that outlier reviews that don't match the mass opinion, i.e., a negative review for a well-rated business, are generally considered less credible or helpful, which means that reviewers that consistently deviate are likely to have recognizable traits rather than being single outlier reviews.These findings offer a solid foundation for characterizing reviewers as generous or critical and show that these differences are not noise but significant features of user behavior.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

Other prior work has examined the contribution of extremity bias to internet reviews and its effects on representativeness. For example, we found the journal article "Online Review Solicitations Reduce Extremity Bias in Online Review Distributions and Increase Their Representativeness" (HÃ¼lya Karaman, 2021). What we can observe here from this journal is that extreme reviews, both extremely positive and extremely negative ones, tend to be more of a reflection of intrinsic reviewer bias rather than the actual reflection of the customers' general experience. Identifying these patterns allows researchers to distinguish better between reviewers who consistently assign extreme ratings and do so on purpose from reviewers who are simply outliers or biased toward certain experiences. This data supports the necessity of distinguishing reviewers based on their tendency to provide unusually high or low ratings. It suggests that harsh and lavish reviewers may engage in systematic behavior for several businesses and categories.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)

To supplement star ratings, the content of written reviews can provide a more insightful picture of the behavior of reviewers. The literature that currently exists in natural language processing (NLP) indicates that text analysis of reviews can reveal sentiment, prevailing themes, and behavioral tendencies of customers. For instance, the GitHub project NLP Sentiment Analysis with Yelp Review Data Set (dailyLi, 2019) employs machine learning models, such as convolutional and recurrent neural networks, for review polarity prediction (positive or negative) and keyword detection driving sentiment. Though this project is targeted at review-level polarity rather than user-level behavior, it demonstrates the potential to combine textual and numeric data for reviewer descriptions. Our research extends this strand of research by examining whether reviewer typesâ€”critical, generous, or neutralâ€”exhibit stable patterns in both ratings and text content, including the proportion of reviews with photos or specific linguistic features.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)

A portion of our team members' prior work, the project Cooking Up Insights, serves as the foundation for the current research. In this project, they analyzed recipe reviews to explore relationships between cooking time and average ratings, examine trends in user behavior across multiple reviews, and measure rating variation. This project provided practical experience in processing user-generated review data, measuring rating deviations, and interpreting behavioral signals, all of which are directly applicable to reviewer type analysis on Google Maps. Building on these methods, our current project systematically identifies reviewer typesâ€”critical, generous, and neutralâ€”and contrasts their behaviors across several dimensions, including rating deviation, review frequency, textual sentiment, and photo usage. While prior research has examined review helpfulness, sentiment analysis, and extremity bias, the literature lacks an analysis of user-level behavioral patterns across multiple reviews and business types. This project seeks to close that gap by combining rating analysis and user behavior metrics, contributing to the literature on online review dynamics and reviewer typologies.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)

- "critical users" has generally longer review length, lower average rating, and more frequent reviews than other groups.
