## Research Question

How do review behaviors, average review length, star rating, review frequency, and geographic diversity, differ among critical, generous, and neutral Google Maps users in California (up to Sep 2021), based on ratings of highly-rated or harshly-rated restaurants?

- `Highly-rated businesses` are defined as locations with at least 100 reviews and an average rating above 4.5, while `harshly-rated restaurants` have at least 100 reviews and an average rating below 3.5.
- Users with at least 100 reviews are considered; `critical users` are those who rated at least two highly-rated restaurants 1 or 2 stars, `generous users` are those who rated at least two harshly-rated restaurants 5 stars, and `neutral users` are all other users.

## Background and Prior Work

Online reviews play a significant role in consumers' decision-making process and a business's reputation, particularly on Google Maps. Users influence potential customers, search engine positioning, and, in some cases, whether a business thrives through ratings in the form of stars, written commentary, and even pictures they post. Although they pervade, there is still much to be discovered about the reviewers themselves, such as how different users behave across multiple reviews and whether certain users consistently deviate from the majority in recognizable ways. For this project, we will identify "critical," `generous," and "neutral" reviewers based on how their ratings compare to the average rating of the establishments they review. We will also determine whether these categories of reviewers exhibit distinct behavioral patterns.

Research has noted that online review distributions are highly polarized, with most users leaving extremely positive or extremely negative reviews, and proportionately fewer in between. This theory confirms that we can successfully categorize and recognize reviewers into at least two categories: critical and generous. Studies such as When Consistency Matters: The Effect of Valence Consistency on Review Helpfulness (Quaschning, Pandelaere & Vermeir, 2014) show that outlier reviews that don't match the mass opinion, i.e., a negative review for a well-rated business, are generally considered less credible or helpful, which means that reviewers that consistently deviate are likely to have recognizable traits rather than being single outlier reviews.These findings offer a solid foundation for characterizing reviewers as generous or critical and show that these differences are not noise but significant features of user behavior.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

Prior research indicates that online review distributions are often polarized, with most users leaving extremely positive or extremely negative reviews and relatively few moderate ones (Quaschning, Pandelaere & Vermeir, 2014). This polarization supports the feasibility of categorizing reviewers into critical and generous types. For instance, Quaschning et al. (2014) found that reviews deviating from the consensus—such as negative reviews of highly rated establishments—are generally perceived as less helpful, suggesting that reviewers who consistently deviate have identifiable behavioral traits rather than being isolated outliers.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)

To supplement star ratings, the content of written reviews can provide a more insightful picture of the behavior of reviewers. The literature that currently exists in natural language processing (NLP) indicates that text analysis of reviews can reveal sentiment, prevailing themes, and behavioral tendencies of customers. For instance, the GitHub project NLP Sentiment Analysis with Yelp Review Data Set (dailyLi, 2019) employs machine learning models, such as convolutional and recurrent neural networks, for review polarity prediction (positive or negative) and keyword detection driving sentiment. Though this project is targeted at review-level polarity rather than user-level behavior, it demonstrates the potential to combine textual and numeric data for reviewer descriptions. Our research extends this strand of research by examining whether reviewer types—critical, generous, or neutral—exhibit stable patterns in both ratings and text content, including the proportion of reviews with photos or specific linguistic features.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)

One of our team members' prior work, Cooking Up Insights (Ventura, 2025), explored user behavior in recipe reviews, examining rating deviations, trends across multiple reviews, and relationships between review content and ratings. This prior experience informs our methodology for processing review data and measuring user-level behavioral metrics. Building on these methods, our current project systematically identifies reviewer types and analyzes differences across rating deviations, review frequency, text sentiment, and photo usage. While prior literature has examined review helpfulness, sentiment, and extremity bias, few studies analyze user-level behavior across multiple establishments. Our project fills this gap, providing a comprehensive characterization of reviewer types on Google Maps.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)

### Loading Dataset

````python
def load_user_reviews():
    critical = pd.read_csv(f"{processed_path}/critical_user_reviews.csv")
    generous = pd.read_csv(f"{processed_path}/generous_user_reviews.csv")
    neutral = pd.read_csv(f"{processed_path}/neutral_user_reviews.csv")

    return critical, generous, neutral

def load_combined_reviews():
    critical, generous, neutral = load_user_reviews()

    critical["group"] = critical["user_id"].apply(lambda x: "critical")
    generous["group"] = generous["user_id"].apply(lambda x: "generous")
    neutral["group"] = neutral["user_id"].apply(lambda x: "neutral")

    combined = pd.concat([critical, neutral, generous], axis=0)

    return combined

def load_stats():
    stats = pd.read_csv(f"{results_path}/stats.csv")
    return stats
````

### Statistics

First, we calculate user statitics as follows:

- `rating_mean`: mean of ratings by a user, representing overall tendency of rating.
  
- `rating_std`: standard eviation of ratings by a user, representing how a user changes their rating for each restaurant.
  
- `num_reviews`: the number of reviews a user rated in total.
  
- `num_reviews_monthly_mean`: mean of the number of reviews a user rated per month, representing how active a user is given how a user is using Google Maps.
  
- `num_reviews_monthly_std`: standard deviation of the number of reviews a user rated per month, representing how much user changes their review frequency by month.
  
- `num_reviews_monthly_max`: maximum number of reviews a user rated per month, representing how much active a user was when a user was most active.
   
- `review_length_mean`: mean of review length in words, representing how long a user tend to write a review for each restaurant on average.
  
- `review_length_std`: standard deviation of review length in words, representing how a user changes length of their reviews for each restaurants.
  
- `num_cities`: the number of cities of restaurants a user has rated, representing how a user is active in terms of their geographical coverage.
  
- `sentiment_mean`: mean of component of nltk's sentiment analysis of user reviews, representing how positive user reviews are on average.

- `sentiment_std`: standard deviation of nltk's sentiment analysis of user reviews, representing how a user changes sentiment of their reviews for each restaurant.
  
- `activity_months`: the number of months a user rated at least one review, representing how long a user has been active.
  
- `activity_duration`: duration in terms of months between a user's first review and last review, representing how long a user has been using Google Maps including months a user was not active.
  
- `activity_rating_diff`: difference between mean of ratings for a user's first 10 reviews and last 10 reviews (last - first), representing how does a user changed rating from when a user started reviewing and a user is currently reviewing. (We may allow overlaps for such 10 reviews.)

- `diversity_mean`: mean of ratios of the number of unique words over the number of words in user reviews, representing how much a user use unique words in their reviews.
  
- `diversity_std`: standard deviation of ratios of the number of unique words over the number of words in user reviews, representing how much a user change word choice for each restaurants.

### Test of Independence

Here, we conducted testing on statsistics we calculated above to see if the difference of such statistics between groups are significant or not.  Overall, we define null hypothesis as

```
The review behaviors of these three users groups are not different.
```

For each user statistic, we have the following steps to conduct a testing:

1. First, we check the statistics follow normal distribution manually.
2. If they does not follow normal distribution, try to find a transformation which transform them into normal distribution by manually checking.
3. Run test and calculate p value for each test.
   - If orignal statistic or tranformed statistic follow normal distirbution conduct two sample t test and Mann Whitney U test.
   - If neither of original statistic nor transformed statistic follow normal distribution, then conduct only Mann Whitney U test.

Note: we checked normality manually because using goodness of fit test requires data to be very close to normal even though two sample t test is robust for larger samples (we have 4155 x 3 = 12465 users). We tried goodness of fit test in `./modules/test_stats` and results are in `./results/distribution.json`.

1. Given plots we have above, we decided to assume `sentiment_std` follows normal distribution.
2. We applied following 4 transformations and reults are inside `./resutls/transform` (we excluded since there are too many plots):
   - `outlier`: just remove outliers. For most statistics, we remove outliers whose z-score is larger than 3.
   - `log`: compute the natural log of statistics
   - `reciprocal`: compute the reciprocal of statistics
   - `sqrt`: compute squared root of statistics
   - `cbrt`: compute cube root of statistics
3. We saved the results as follows:
   - `raw_normal`: indicate if raw data is normal.
   - `transformed_normal`: indicates if transformed data is normal.
   - `test_transform`: indicates the method for transformation which could transform data into normal.

Notes:
- The implementation for transformation and removing outliers are in `./modules/utils/plot_utils.py`
- The implementation for plotting transform is embedded inside `DistributionVisualizer` class from `./modules/visualization/visualize_stats_distribution.py`.

````python
dists = {
    "activity_duration": {
        "normal": 0,
        "transformed_normal": 1,
        "test_transform": "sqrt"
    },
    "activity_months": {
        "normal": 0,
        "transformed_normal": 0
    },
    "activity_rating_diff": {
        "normal": 1
    },
    "diversity_mean": {
        "normal": 0,
        "transformed_normal": 0
    },
    "diversity_std": {
        "normal": 1
    },
    "num_cities": {
        "normal": 0,
        "transformed_normal": 0
    },
    "num_reviews_monthly_max": {
        "normal": 0,
        "transformed_normal": 0
    },
    "num_reviews_monthly_mean": {
        "normal": 0,
        "transformed_normal": 1,
        "test_transform": "log"
    },
    "num_reviews_monthly_std": {
        "normal": 0,
        "transformed_normal": 1,
        "test_transform": "log"
    },
    "num_reviews": {
        "normal": 0,
        "transformed_normal": 0
    },
    "rating_mean": {
        "normal": 0,
        "transformed_normal": 0
    },
    "rating_std": {
        "normal": 0,
        "transformed_normal": 1,
        "test_transform": "outlier"
    },
    "review_length_mean": {
        "normal": 0,
        "transformed_normal": 1,
        "test_transform": "cbrt"
    },
    "review_length_std": {
        "normal": 0,
        "transformed_normal": 1,
        "test_transform": "log"
    },
    "sentiment_mean": {
        "normal": 1
    },
    "sentiment_std": {
        "normal": 1
    }
}
````

Then, we define a tester class which test each statsistic with procedures we described above.

````python
class StatisticsTester():
    def __init__(self, stats):
        self.stats = stats
        self.groups = np.unique(self.stats["group"])

        self.transforms = ["outlier", "log", "reciprocal", "sqrt", "cbrt"]

        self.p_thresh = 0.01

    def goodness_of_fit(self, values, name):
        p_values = []
        for group in self.groups:
            group_stats = values[values["group"] == group][name]
            params = {"loc": np.mean(group_stats), "scale": np.std(group_stats)}

            res = scipy.stats.goodness_of_fit(scipy.stats.norm, group_stats, known_params=params)
            p_values.append(res.pvalue)

        p_value = min(p_values)

        return p_value

    def two_sample_t_test(self, group_0, group_1):
        res = scipy.stats.ttest_ind(group_0, group_1, equal_var=False)
        return res.pvalue, res.statistic

    def mann_whitney_u_test(self, group_0, group_1):
        res = scipy.stats.mannwhitneyu(group_0, group_1)
        return res.pvalue, res.statistic

    def test_raw_normal(self):
        with open(f"{results_path}/stats.txt", "r") as f:
            stats_list = f.read().split()

        results = {}
        for stats in stats_list:
            print(f"Testing raw normal for {stats}")
            values = self.stats[["group", stats]]
            p_value = self.goodness_of_fit(values, stats)
            results[stats] = {"raw_normal": int(p_value >= self.p_thresh), "raw_normal_p": p_value}

        with open(f"{results_path}/distribution.json", "w") as f:
            json.dump(results, f)

    def test_transform_normal(self):
        with open(f"{results_path}/distribution.json", "r") as f:
            results = json.load(f)

        old_results = results
        for key, values in old_results.items():
            if results[key]["raw_normal"]:
                continue

            if "transform_normal" in results[key].keys():
                continue

            for transform in self.transforms:
                print(f"Testing raw normal for {key} with {transform}")

                values = self.stats[["group", key]].dropna()
                values.iloc[:, 1] = apply_transform(values[key].values, transform=transform)

                p_value = self.goodness_of_fit(values, key)
                if p_value >= self.p_thresh:
                    results[key]["transform_normal"] = 1
                    results[key]["transform_normal_p"] = float(p_value)
                    results[key]["transform"] = transform

            if "transform_normal" not in results[key].keys():
                results[key]["transform_normal"] = 0

            with open(f"{results_path}/distribution.json", "w") as f:
                json.dump(results, f)

    def independence_tests(self, dists):
        pairs = [(self.groups[i], self.groups[j]) for i in range(3) for j in range(i+1, 3)]
        results = []

        for key, dist in dists.items():
            values = self.stats[["group", key]].dropna()

            if dist["normal"] == 1:
                t_test = True
            elif dist["transformed_normal"] == 1:
                t_test = True
                values[key] = apply_transform(values[key].values, transform=dist["test_transform"])
            else:
                t_test = False

            for (group_0, group_1) in pairs:
                values_0 = values[values["group"] == group_0][key].dropna().values
                values_1 = values[values["group"] == group_1][key].dropna().values

                if t_test:
                    t_test_p, t_test_stat = self.two_sample_t_test(values_0, values_1)
                    log_t_test_p = -np.where(t_test_p == 0, -500, np.log10(t_test_p))
                else:
                    t_test_p = None
                    log_t_test_p = None
                    t_test_stat = None

                u_test_p, u_test_stat = self.mann_whitney_u_test(values_0, values_1)
                log_u_test_p = -np.where(u_test_p == 0, -500, np.log10(u_test_p))

                results.append((key, f"{group_0}_{group_1}",
                                t_test_p, u_test_p,
                                t_test_stat, u_test_stat,
                                log_t_test_p, log_u_test_p
                ))

        results = pd.DataFrame(results, columns=[
            "name", "pair", "t_test_p", "u_test_p", "t_test_stat", "u_test_stat", "log_t_test_p", "log_u_test_p"
        ])
        print(results)

        os.makedirs(f"{results_path}/tests", exist_ok=True)
        results.to_csv(f"{results_path}/tests/results.csv", index=False)

        return results

    def plot_test_results(self, results, show):
        _, axes = plt.subplots(nrows=1, ncols=2, figsize=(25, 10))

        sns.set_style("whitegrid")
        plt.suptitle(f"Comparisons of p-values of test results (red dot line corresponds to 0.01)")

        axes[0].set_title("Comparison of log of p values for two sample t test.\n(if negative log of p value is 0, then it is replaced as 500)")
        axes[0].set_xlabel("Negative log of p value")
        axes[0].set_ylabel("Statistic")
        axes[0].axvline(-np.log(self.p_thresh), color="red", linestyle="dashed")
        sns.barplot(results, y="name", x="log_t_test_p", hue="pair", ax=axes[0], alpha=0.8)

        axes[1].set_title("Comparison of p values for Mann Whitney U test.")
        axes[1].set_xlabel("Negative log of p value")
        axes[1].set_ylabel("Statistic")
        axes[1].axvline(-np.log(self.p_thresh), color="red", linestyle="dashed")
        sns.barplot(results, y="name", x="log_u_test_p", hue="pair", ax=axes[1], alpha=0.8)

        plt.tight_layout()
        if show:
            plt.show()
        else:
            plt.savefig(f"{results_path}/tests/p_comparison.png")
            plt.close()

        _, axes = plt.subplots(nrows=1, ncols=2, figsize=(25, 10))
        plt.suptitle(f"Comparisons of p-values of test results (red dot line corresponds to critical value for 0.01)")

        axes[0].set_title("Comparison of test statistics for two sample t test.")
        axes[0].set_xlabel("Value of test statistic")
        axes[0].set_ylabel("Statistic")
        axes[0].axvline(-2.57644, color="red", linestyle="dashed")
        axes[0].axvline(2.57644, color="red", linestyle="dashed")
        sns.barplot(results, y="name", x="t_test_stat", hue="pair", ax=axes[0], alpha=0.8)

        axes[1].set_title("Comparison of test statistics for Mann Whitney U test.")
        axes[1].set_xlabel("Value of test statistic")
        axes[1].set_ylabel("Statistic")
        sns.barplot(results, y="name", x="u_test_stat", hue="pair", ax=axes[1], alpha=0.8)

        plt.tight_layout()

        if show:
            plt.show()
        else:
            plt.savefig(f"{results_path}/tests/stat_comparison.png")
            plt.close()

    def test_statistics(self, dists, show=False):
        if not os.path.exists(f"{results_path}/distribution.json"):
            self.test_raw_normal()
            self.test_transform_normal()

        results = self.independence_tests(dists)
        self.plot_test_results(results, show)
````

These graphs show the p-values and test statistics for each test we conducted. Since p values tend to be way too small to analyze we used negative log10 function to scale the values. Some tests showed 0 as p-value which caused error in computing log, so I capped the value as 500. Moreover, we used 0.01 as an threshold to decided if the results are significant or not.

First graph shows that in most user statistics we found that there are siginificant difference between user groups. ### Dimensionality Reduction

- `t-SNE`: t-distributed stochastic neighbor embedding is a non linear dimensionality reduction tecnique.

### Classifiers

In addition to dimensionality reduction we applied above, we decided to train a simple classifier on user statistics to see if classifiers can classify user's group based on user statistics above and how they are separable numerically. If difference of review behavior are significant enought, we assume that classifier can classify them accurately.

````python
def classify_users(stats, models, show=False):
    vars = list(stats.columns)[2:]
    values = stats[vars].values
    labels = stats["group"].values

    x_train, x_test, y_train, y_test = train_test_split(values, labels, test_size=0.1, random_state=0)

    sns.set_style("whitegrid")
    _, axes = plt.subplots(nrows=2, ncols=2, figsize=(15, 10))
    plt.suptitle("Confusion matrices for classifiers on test dataset.")

    accuracies = []
    for i in range(len(models)):
        name, model = models[i]
        print(f"Classifying users with {name}")

        model = model.fit(x_train, y_train)

        ax = axes.flatten()[i]
        ax.grid(False)
        ConfusionMatrixDisplay.from_estimator(model, x_test, y_test, ax=ax, cmap="GnBu")
        ax.set_title(name)

        y_train_pred = model.predict(x_train)
        y_test_pred = model.predict(x_test)

        train_accuracy = np.sum(y_train_pred == y_train) / y_train.shape[0]
        test_accuracy = np.sum(y_test_pred == y_test) / y_test.shape[0]

        accuracies.append((name, train_accuracy, "train"))
        accuracies.append((name, test_accuracy, "test"))

    plt.tight_layout()

    if show:
        plt.show()
    else:
        os.makedirs(f"{results_path}/classification", exist_ok=True)
        plt.savefig(f"{results_path}/classification/confusion_matrix.png")
        plt.close()

    df = pd.DataFrame(accuracies, columns=["model", "accuracy", "split"])
    print(df)

    sns.set_style("whitegrid")
    plt.figure(figsize=(15, 10))
    plt.title("Comparison of train and test accuracies with different models.")
    sns.barplot(df, x="model", y="accuracy", hue="split", alpha=0.8)

    plt.xlabel("Models")
    plt.xlabel("Accuracy")

    plt.ylim(0.5, 0.8)
    plt.tight_layout()

    if show:
        plt.show()
    else:
        plt.savefig(f"{results_path}/classification/accuracies.png")
        plt.close()
````

````python
models = [
    ("Logisitc Regression", LogisticRegression(max_iter=10000)),
    ("Support Vector Machine", SVC(kernel="rbf")),
    ("Random Forest", RandomForestClassifier(max_depth=5)),
    ("Ada Boost", AdaBoostClassifier())
]

classify_users(stats, models, show=True)
````

From the first figure of confusion matrices, we can see models can reasonably classify most users but fail to classify correctly for some users. The models are most confused with difference between `generous` and `neutral` users which also showed less significant in testing above.

Since we did not include features which tells the criteria or definitions for groups (how users rated harshly-rated or highly-rated restaurants), classifiers had to infer such conditions just based on other user statistics we calculated above. Overall, we got around 70% accuracy overall. Moreover, suprisingly, simplest model, linear regression showed the best performance, which tells us that users are `linearly` separable in high dimensional user statistics spcae to some extent. This also supports our testing that the distributions of review behaviors of users are not the same.

Generous users use the least variety, aligning with their shorter, repetitive style. Paired analyses show that review length and rating mean together strongly predict group membership: long reviews with low ratings indicate critical users, short reviews with high ratings indicate generous users, and neutral users appear between these extremes. ### Bigger Picture

Taken as a whole, the results reveal that reviewer behavior is not random but falls into stable, meaningful archetypes. Critical users emerge as strict evaluators who write long, detailed, varied reviews; generous users emerge as consistent high raters who write frequent but short and simple reviews; neutral users serve as the statistical middle ground. This means that online review ecosystems, especially platforms like Google Maps, are shaped less by an “average user” and more by the distinctive behaviors of these two extreme groups. Their influence is amplified by their activity levels, broader geographic engagement, and longer reviewing histories.
These patterns carry significant implications for the interpretation and design of review platforms. Generous users heavily shape rating averages due to their volume and consistent 5-star behavior, but they do not shape the narrative understanding of a business because their language is simple and brief. Critical users, by contrast, exert outsized influence on the textual narrative, providing detailed feedback and a broader emotional and descriptive range that affects how future customers interpret business quality. Neutral users anchor the system with balanced, representative feedback but have limited influence on the larger rating distribution due to their lower activity.
Overall, your study demonstrates that reviewer identity matters as much as business quality in shaping online ratings. The stability of these behavioral clusters, across text length, vocabulary diversity, sentiment, geography, and temporal engagement suggests deeply rooted reviewer identities rather than random variations. Understanding these identities is essential for building fairer recommendation systems, correcting for systematic rating biases, and interpreting business reputations more accurately. In short, online review platforms are ecosystems driven not merely by what is being rated but by who is doing the rating and how they choose to speak.

### Contradictory Results

Our exploratory data analysis presented several contradictory results within our review data. For example, the sentiment scores we discovered did not always align with the ratings. Some of the reviews that were highly rated only had neutral sentiment when we were expecting positive sentiment. Furthermore, some low-rated reviews showed positive sentiment, which was also unexpected. We suspect that this mismatch is due to the limitations of sentiment models, as we infer they may struggle with reviews that have mixed tones, sarcasm, or writing where readers infer emotions rather than highly charged language. Additionally, while analyzing review length, longer reviews did not show a consistent negative or positive sentiment. This contradicts our initial hypothesis, and we believe that it is most likely that extended text contains both praise and negative feedback. Ultimately, our results suggest that some inconsistencies do exist and we believe that it is because user ratings, textual sentiment, and review length illustrate different reviewing experience aspects. Thus, the three features do not always move in sync and therefore present unexpected patterns.

### Limitations

Nevertheless, there were also multiple limitations present in the dataset that prevented us from drawing all the conclusions we wanted. One limitation was that our data suffers from selection bias, as the individuals who are frequent reviewers or reviewers who tend to leave long reviews are not representative of all reviewers. This selection process that we underwent when examining review length and such may have shaped observed trends in a cherry-picking manner. Also, our data lacked demographic information, and product metadata restricted our ability to fully understand the context as well as make adjustments for confounding factors. Lastly, sentiment scores may have introduced noise as the models may have misinterpreted mixed opinions. If that is the case, it would reflect the poor ability of the sentiment models as they limit us from examining how sentiment truly varies across different reviews. Altogether, these limitations we encountered cause us to be slightly uncertain about our interpretation of our findings.
